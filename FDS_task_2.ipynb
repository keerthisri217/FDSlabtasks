```python
# ==========================================
# AUTOMOBILE DATASET ANALYSIS
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv('Automobile_data.csv')

# Display Dataset Information
print("First 5 Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())

# Convert required columns to numeric
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df['engine-size'] = pd.to_numeric(df['engine-size'], errors='coerce')
df['city-mpg'] = pd.to_numeric(df['city-mpg'], errors='coerce')

sns.set_style("whitegrid")

# ==========================================
# 1. Count Plot - Body Style
# ==========================================
plt.figure(figsize=(8,5))
sns.countplot(x='body-style', data=df)
plt.title("Number of Cars by Body Style")
plt.xticks(rotation=45)
plt.show()

# ==========================================
# 2. Bar Plot - Average Price by Fuel Type
# ==========================================
plt.figure(figsize=(7,5))
sns.barplot(x='fuel-type', y='price', data=df)
plt.title("Average Price by Fuel Type")
plt.show()

# ==========================================
# 3. Histogram - Price Distribution
# ==========================================
plt.figure(figsize=(8,5))
sns.histplot(df['price'].dropna(), bins=20, kde=True)
plt.title("Price Distribution")
plt.show()

# ==========================================
# 4. Box Plot - Price vs Fuel Type
# ==========================================
plt.figure(figsize=(7,5))
sns.boxplot(x='fuel-type', y='price', data=df)
plt.title("Price by Fuel Type")
plt.show()

# ==========================================
# 5. Scatter Plot - Horsepower vs Price
# ==========================================
plt.figure(figsize=(8,5))
sns.scatterplot(x='horsepower', y='price', data=df)
plt.title("Horsepower vs Price")
plt.show()

# ==========================================
# 6. Pie Chart - Fuel Type
# ==========================================
plt.figure(figsize=(6,6))
df['fuel-type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Fuel Type Distribution")
plt.ylabel("")
plt.show()

# ==========================================
# 7. Correlation Heatmap
# ==========================================
plt.figure(figsize=(10,8))
numeric_df = df.select_dtypes(include=['number'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# ==========================================
# 8. Count Plot - Drive Wheels
# ==========================================
plt.figure(figsize=(7,5))
sns.countplot(x='drive-wheels', data=df)
plt.title("Drive Wheels Count")
plt.show()

# ==========================================
# 9. Violin Plot - Price vs Number of Doors
# ==========================================
plt.figure(figsize=(7,5))
sns.violinplot(x='num-of-doors', y='price', data=df)
plt.title("Price by Number of Doors")
plt.show()

# ==========================================
# 10. Pair Plot
# ==========================================
cols = ['price', 'horsepower', 'engine-size', 'city-mpg']
sns.pairplot(df[cols].dropna())
plt.show()
```
